## Quick look at the data

The starter dataset contains **30,000 observations and 30,000 unique content IDs**, so each observation corresponds to a unique content item in this starter dataset.

There are **32 unique clients** represented in the dataset.

The `trend_direction` column contains **16,262 observations classified as `down`, which is approximately 54.21% of the dataset**.

The median number of 90-day impressions is **731**, while the mean is approximately **5,200**. The large difference between the median and mean suggests that impression volume is unevenly distributed across content items.

These initial numbers make the refresh/content opportunity problem worth investigating further. However, the downward-trend label by itself does not establish that a page should be refreshed or that a refresh would improve its performance.


```python
down_count = (df["trend_direction"] == "down").sum()
down_percentage = down_count / len(df) * 100

print(f"Pages with downward trend: {down_count:,}")
print(f"Percentage with downward trend: {down_percentage:.2f}%")
```
